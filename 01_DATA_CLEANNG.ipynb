```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plit
import seaborn as sns

print("Retail sales data analytics project started ")
print("All libraries im[ported successfully.")
```

```python
data = {
    "Order_ID": ["101", "102", "103", "104", "105"],
    "Customer_Age": [25, 30, None, 45, 200],
    "Gender": ["Male", "female", "M", "FEMALE", "Male"],
    "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone"],
    "Sales": [50000, 20000, 15000, None, 25000],
    "Order_Date": ["2026-01-05", "05/02/2026",
                   "2026-03-10", "15-03-2026", "2026-04-01"]
}

df = pd.DataFrame(data)

df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

print("Dataset created successfully!")
display(df)
```

```python
# DATA QUALITY REPORT

print("===== DATA QUALITY REPORT =====")

# 1. Dataset shape
print("\nDataset Shape:", df.shape)

# 2. Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 3. Duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# 4. Data types
print("\nData Types:")
print(df.dtypes)

# 5. Numerical summary
print("\nNumerical Summary:")
display(df.describe())

# 6. Age anomalies
print("\nInvalid Customer Ages:")
print(df[(df["Customer_Age"] < 0) |
         (df["Customer_Age"] > 100)])

# 7. Negative sales
print("\nNegative Sales:")
print(df[df["Sales"] < 0])

print("\n===== REPORT COMPLETED =====")
```

1. Customer_Age: Missing value will be replaced using median because median is less affected by extreme values.
2. Sales: Missing value will be replaced using median to avoid the influence of unusually high sales values.
3.Duplicate records will be removed before calculating final imputation values

```python
# STEP 4: HANDLE MISSING VALUES

# Remove duplicate rows
df = df.drop_duplicates()

print("Duplicates removed successfully!")
print("Remaining rows:", len(df))

# Replace invalid customer age with missing value
df.loc[df["Customer_Age"] > 100, "Customer_Age"] = np.nan

# Fill missing age using median
df["Customer_Age"] = df["Customer_Age"].fillna(
    df["Customer_Age"].median()
)

# Fill missing sales using median
df["Sales"] = df["Sales"].fillna(
    df["Sales"].median()
)

print("\nMissing values after cleaning:")
print(df.isnull().sum())

display(df)
```

1.Gender: Inconsistent values such as "M", "F", "male" and "FEMALE" will be standardised into "Male" and "Female".
2.Order_Date: Different date formats will be converted into a consistent datetime format.
3.Order_ID: IDs will be stored as strings because they are identifiers, not numerical values.
4.Sales: Monetary values will be stored as float.

```python
# STEP 5: DATA STANDARDISATION

# 1. Standardise Gender
gender_mapping = {
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female"
}

df["Gender"] = (
    df["Gender"].str.strip().str.lower()
    .map(gender_mapping)
)

# 2. Convert dates into datetime format
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

# 3. Correct data types
df["Order_ID"] = df["Order_ID"].astype(str)
df["Customer_Age"] = df["Customer_Age"].astype(int)
df["Sales"] = df["Sales"].astype(float)

# Display cleaned data
print("Data standardisation completed!")

print("\nUpdated Data Types:")
print(df.dtypes)

print("\nStandardised Gender Values:")
print(df["Gender"].unique())

display(df)
```

The Interquartile Range (IQR) method is used to identify outliers in numerical columns.
Customer_Age: Outliers will be identified using the IQR method.
Sales: Unusually high or low sales values will be identified.
Decision: Valid observations will be retained, while extreme sales values will be capped at the IQR upper or lower boundary.
Capping is preferred over row deletion to preserve useful records.

```python
# STEP 6: OUTLIER DETECTION USING IQR

# Add a synthetic test anomaly
df.loc[df["Order_ID"] == "105", "Sales"] = 999999.0

# Function to detect outliers
def detect_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower) |
        (data[column] > upper)
    ]

    print("\nColumn:", column)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
    print("Outliers Found:", len(outliers))

    return lower, upper

# Check Customer Age
age_lower, age_upper = detect_outliers(
    df, "Customer_Age"
)

# Check Sales
sales_lower, sales_upper = detect_outliers(
    df, "Sales"
)

# Cap extreme Sales values
df["Sales"] = df["Sales"].clip(
    lower=sales_lower,
    upper=sales_upper
)

print("\nOutlier treatment completed!")
display(df)
```

This section compares the dataset before and after cleaning. It evaluates missing values, duplicate records, row count and data type accuracy to measure the effectiveness of the cleaning process.
